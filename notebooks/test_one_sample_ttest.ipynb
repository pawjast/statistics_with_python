# One sample t-test

* Scipy docs: [link](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.ttest_1samp.html)

## Imports

````python
import scipy.stats as stats
import numpy as np
````

## Scenario 1: two sided test

**Context**

A tech team redesigns a checkout page. The historical average load time is $\mu_0 = 2.0$ seconds. They want to test whether the new page load time differs from the baseline (either faster or slower).

**Hypothesis**

* $H_0: \mu = 2.0$ (mean load time equals 2.0s)
* $H_1: \mu \neq 2.0$ (mean load time differs from 2.0s)
* Significance level alpha: 0.05 (5%)

**Sample info**

* Sample Data: $n = 25$ page loads
* Sample mean $\bar{x} = 2.18$ seconds
* Sample standard deviation $s = 0.45$ seconds.
* Degrees of Freedom: $df = n - 1 = 25 - 1 = 24$

### Inputs

````python
POPULATION_MEAN = 2.0 # seconds
SAMPLE_MEAN = 2.18  # seconds
SAMPLE_STD_DEV = 2.18
NR_OF_OBSERVATIONS = 25
ALPHA = 0.05
````

### Data: page loads

Generating synthetic data matching our sample stats (mean=2.18, s=0.45, n=25)

````python
np.random.seed(13)

data = np.random.normal(
    loc=SAMPLE_MEAN,
    scale=SAMPLE_STD_DEV,
    size=NR_OF_OBSERVATIONS
)

data
````

### Test

````python
test_results = stats.ttest_1samp(
    data,
    popmean=POPULATION_MEAN
)
````

````python
test_results
````

````python
# Unpack results to individual variables
t_stat = test_results.statistic  # test statistics
p_val = test_results.pvalue
dof = test_results.df
````

### Conclusion using p-value and alpha

````python
if p_val < ALPHA:
    print("Reject H0")
else:
    print("Fail to reject H0")
````

### Conclucsion using critical value

Two sided test:

-t_critical < t_statistics < t_critical

````python
# Degrees of freedom
dof = NR_OF_OBSERVATIONS - 1
dof
````

````python
# Two sided test

t_critical_left = stats.t.ppf(
    q=ALPHA / 2,  # at 2.5%
    df=dof
)
t_critical_right = stats.t.ppf(
    q=1 - ALPHA / 2,  # at 97.5%
    df=dof
)

print(f"critical value - low: {t_critical_left:.4f}")
print(f"critical value - high: {t_critical_right:.4f}")
````

Decision rule visualized:

```
   2.5%         95%          2.5%
<--------|----------------|-------->
       -2.064          +2.064
```

The `R` denotes the null hypothesis rejection area:

```
   2.5%         95%          2.5%
<RRRRRRRR|----------------|RRRRRRRR>
       -2.064          +2.064
```

````python
if t_stat < t_critical_left or t_stat > t_critical_right:
    print("Reject H0")
else:
    print("Fail to reject H0")
````
