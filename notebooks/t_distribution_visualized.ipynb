# Understanding t-distribution with Python

## Imports

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

## t-ditribution Intro

* It's a distribution that has a bell-curved, normal shape
* But it's got heavier tails
* The exact shape is determined by the degrees of freedom `dof`
  * `dof = n - 1` where `n` is a number of observations in the sample
  * the higher the `dof` the more the shape of *t-distribution* converges to the normal distribution (more observations == less uncertainty)

## Going from random variable x to t-value

The random variable can be anything that represents a continous measurement. E.g. a temperature.

To calculate the t value, we need a sample of observations for which we can calculate the mean, the standard deviation:

$$
t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}
$$

Where:

* $\bar{x}$: sample mean
* $\mu_0$: Hypothesized population mean under $H_0$
* $s$: sample standard deviation
* $n$: sample size
* $t$: test statistic

## PDF - probabilty density function

PDF is a mathematical function that decribes how probability is distributed across the possible values of a continuous random variable.

PDF mathematically define the shape, spread, and central tendency of continuous theoretical distribution.

PDF describes the probability density.

* The higher the curve at a particular value, the more likely observations are to occur near that value.

The actual probability is the area under the curve (within a specific interval).

As we've got a continous variable, the probability for any specific point is exactly 0.

The total area under the PDF curve equals 1.

In other words, the PDF will tell you:

> How tall is the curve at any given x?



```python
t = np.linspace(-6, 6, 200)  # range of t statistic values
dofs = [1, 3, 5, 10, 20]
```

```python
fig, ax = plt.subplots(
    figsize=(12,6),
    constrained_layout=True
)

ax.set_title("T-distributions vs Normal")

ax.plot(
    x,
    stats.norm.pdf(x),
    linestyle="--",
    label="Normal"
)

for dof in dofs:
    ax.plot(
        x,
        stats.t.pdf(x, df=dof),
        label=f"dof={dof}"
    )

ax.set_xlabel("t value")
ax.set_ylabel("Probability density f(t)")

ax.grid(
    color="lightcyan",
    lw=1.25,
)
ax.set_axisbelow(True)
ax.legend()


ax.text(
    0.01, 0.88,
    s="Fewer samples == lower dof\nLower dof == heavier tails\nHigher dof = t-distribtion approaches normal",
    fontsize=12,
    color="red",
    fontweight="bold",
    transform=ax.transAxes
);
```
